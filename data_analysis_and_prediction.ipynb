<p style="padding:50px;
          color:#5C4A2A;"><strong>Why Is Diabetes Called a Silent Killer?</strong><br>
Diabetes can develop for a long time without causing obvious symptoms.
Over time, high blood sugar can damage blood vessels and nerves throughout the body.
This damage may affect the heart, kidneys, eyes, and nervous system.
That is why early detection and proper blood sugar control are so important.

</p>
</div>

<div style="color:#5C4A2A;
           display:fill;
           border-radius:100px;
           background:#EBD9B5;
           font-size:120%;
           font-family:Verdana;
           letter-spacing:0.5px">

<p style="padding:50px;
          color:#5C4A2A;"><strong>Is Diabetes Caused Only by Eating Too Much Sugar?</strong><br>
No. Eating too much sugar alone is not the direct cause of every type of diabetes.
Type 2 diabetes usually results from a combination of genetics, insulin resistance, excess weight, and lifestyle factors.
Type 1 diabetes is an autoimmune disease and is not caused by eating sweets.
Therefore, diabetes is much more complex than simply “eating too much sugar.”

</p>
</div>

<div style="color:#5C4A2A;
           display:fill;
           border-radius:100px;
           background:#EBD9B5;
           font-size:120%;
           font-family:Verdana;
           letter-spacing:0.5px">

<p style="padding:50px;
          color:#5C4A2A;"><strong>Why Can’t the Body Properly Control Blood Sugar in Diabetes?</strong><br>
Insulin is a hormone that helps glucose move from the blood into cells, where it can be used for energy.
In type 1 diabetes, the body produces little or no insulin.
In type 2 diabetes, cells become resistant to insulin and its effectiveness decreases.
As a result, more glucose remains in the bloodstream, causing blood sugar levels to rise.

</p>
</div>

<div style="color:#5C4A2A;
           display:fill;
           border-radius:100px;
           background:#EBD9B5;
           font-size:120%;
           font-family:Verdana;
           letter-spacing:0.5px">

<p style="padding:50px;
          color:#5C4A2A;"><strong>Is Managing Diabetes Only About Lowering Blood Sugar?</strong><br>
No. Diabetes management involves much more than controlling glucose levels.
Blood pressure, cholesterol, body weight, nutrition, physical activity, and overall health are also important.
Managing these factors can reduce the risk of complications affecting the heart, kidneys, eyes, and nerves.
That is why diabetes management requires a comprehensive and long-term approach.

</p>
</div>

If a column has moderate importance and around 50% of its values are missing, we should first check the database to see if the missing information can be recovered. If the information is not available in the database, we should remove the column. However, if a column is highly important and its source is not available in the database, we should remove the affected rows instead.
In this dataset, I decided to remove the columns because I don’t have access to another database to recover the missing values, and these columns have moderate importance. So, I removed them.

</p>

```python
count_chol_normal = 0
count_chol_warning = 0
count_chol_dangerous = 0

for count in df["chol"]:

    if (count < 200):

        count_chol_normal += 1

    elif (200 <= count <= 239):

        count_chol_warning += 1

    elif (count >= 240):

        count_chol_dangerous += 1

        


print(f"Number of cholesterol data in the lower range (<200) is: {count_chol_normal}")
print(50 * "==")
print(f"Number of cholesterol data in the range between (200 to 239) is: {count_chol_warning}")
print(50 * "==")
print(f"Number of cholesterol data in the range greater than (>240) is: {count_chol_dangerous}")
```

```python
count_lower = 0
count_normal = 0
count_prediabetes = 0
count_diabetes = 0

for count in df["stab.glu"]:

    if (count < 70):

        count_lower += 1

    elif (70 <= count <= 100):

        count_normal += 1

    elif (100 <= count <= 125):

        count_prediabetes += 1

    elif (count >= 125):

        count_diabetes += 1
        

print(f"Number of data less than 70 is: {count_lower}")
print(50 * "==")
print(f"Number of data between 70 and 100 is: {count_normal}")
print(50 * "==")
print(f"Number of data between 100 and 125 is: {count_prediabetes}")
print(50 * "==")
print(f"The number of data is more than 125 is: {count_diabetes}")
```

```python
# ==============================================================================================
count_fbs_female_lower_loc = df[(df["stab.glu"] < 70) & (df["gender"] == "female")][["location"]]

group_fbs_female_lower = count_fbs_female_lower_loc.groupby("location").size().reset_index(name="count")

# ===========================================================================================================
count_fbs_female_normal_loc = df[(df["stab.glu"].between(70, 100) & (df["gender"] == "female"))][["location"]]

group_fbs_female_normal = count_fbs_female_normal_loc.groupby("location").size().reset_index(name="count")

# ====================================================================================================================
count_fbs_female_prediabetes_loc = df[(df["stab.glu"].between(100.1, 125) & (df["gender"] == "female"))][["location"]]

group_fbs_female_prediabetes = count_fbs_female_prediabetes_loc.groupby("location").size().reset_index(name="count")

# ====================================================================================================
count_fbs_female_diabetes_loc = df[(df["stab.glu"] >= 125) & (df["gender"] == "female")][["location"]]

group_fbs_female_diabetes = count_fbs_female_diabetes_loc.groupby("location").size().reset_index(name="count")





print(f"show count fbs female lower location: {group_fbs_female_lower}")
print(50 * "==")
print(f"show count fbs female normal location: {group_fbs_female_normal}")
print(50 * "==")
print(f"show count fbs female prediabetes location: {group_fbs_female_prediabetes}")
print(50 * "==")
print(f"show count fbs female diabetes location: {group_fbs_female_diabetes}")
```

```python
# subplots 
fig = make_subplots(rows=4, cols=2,
                   specs=[[{"type" : "xy"}, {"type" : "domain"}],
                          [{"type" : "xy"}, {"type" : "domain"}],
                          [{"type" : "xy"}, {"type" : "domain"}],
                          [{"type" : "xy"}, {"type" : "domain"}]],
                   column_widths=[2, 2],
                   horizontal_spacing=0,
                   vertical_spacing=0.15,
                   subplot_titles=["lower" ,"lower" ,"normal", "normal", "prediabetes", "prediabetes", "diabetes", "diabetes"])




loc_lower = group_fbs_female_lower["location"]

count_lower = group_fbs_female_lower["count"]

# ===============================================

loc_normal = group_fbs_female_normal["location"]

count_normal = group_fbs_female_normal["count"]

# =================================================

loc_prediabetes = group_fbs_female_prediabetes["location"]

count_prediabetes = group_fbs_female_prediabetes["count"]

# ==================================================

loc_diabetes = group_fbs_female_diabetes["location"]

count_diabetes = group_fbs_female_diabetes["count"]



# add traces
fig.add_trace(go.Bar(x=loc_lower, y=count_lower, text=count_lower), row=1, col=1)
fig.add_trace(go.Bar(x=loc_normal, y=count_normal, text=count_normal), row=2, col=1)
fig.add_trace(go.Bar(x=loc_prediabetes, y=count_prediabetes, text=count_prediabetes), row=3, col=1)
fig.add_trace(go.Bar(x=loc_diabetes, y=count_diabetes, text=count_diabetes), row=4, col=1)




fig.add_trace(go.Pie(labels=loc_lower, values=count_lower, textfont_size=20,
                     textposition='inside', textinfo='percent+label'), row=1, col=2)

fig.add_trace(go.Pie(labels=loc_normal, values=count_normal, textfont_size=20,
                     textposition='inside', textinfo='percent+label'), row=2, col=2)

fig.add_trace(go.Pie(labels=loc_prediabetes, values=count_prediabetes, textfont_size=20,
                     textposition='inside', textinfo='percent+label'), row=3, col=2)

fig.add_trace(go.Pie(labels=loc_diabetes, values=count_diabetes, 
                     textfont_size=20, textposition='inside', textinfo='percent+label'), row=4, col=2)




# Update xaxis properties
fig.update_xaxes(title_text="location", title_font={"size" : 15}, title_standoff=15)

# Update xaxis properties
fig.update_yaxes(title_text="count", title_font={"size" : 15}, title_standoff=15)




# chart settings
fig.update_layout(
    width=1100,
    height=1300,
    font_size=15,
    showlegend=False,



    # add title
    title="Investigating the number and percentage of women in Louisa and Buckingham positions with fasting blood sugar in lower, normal, prediabetic, and diabetic states",
    title_font_size=12,
    title_font_color="grey",
    title_font_weight='bold',
    title_x=0.5,
    title_y=0.98
)





fig.show()
```

This will be examined in terms of low and normal levels, prediabetes and diabetes.
</div>

```python
count_normal = 0
count_in_danger = 0
count_dangerous = 0

for count in df["hdl"]:

    if (count >= 60):

        count_normal += 1

    elif (40 <= count <= 59):

        count_in_danger += 1

    elif (count <= 40):

        count_dangerous += 1


print(f"Number of data in normal (HDL >= 60) is: {count_normal}")
print(50 * "==")
print(f"Number of data in in_danger(40 <= HDL <= 59) is: {count_in_danger}")
print(50 * "==")
print(f"Number of data in dangerous (HDL <= 40) is: {count_dangerous}")
```

Obtaining the number of systolic blood pressure data in the range of less than (120) and between (120 - 129) and between (130 - 139) and greater than (140) and greater than (180)

```python
count_systolic_normal = 0
count_systolic_increased = 0
count_systolic_more_stage1 = 0
count_systolic_more_stage2 = 0
count_systolic_dangerous = 0


for count in df["bp.1s"]:

    if (count < 120):

        count_systolic_normal += 1

    elif (120 <= count <= 129):

        count_systolic_increased += 1

    elif (130 <= count <= 139):

        count_systolic_more_stage1 += 1

    elif (140 <= count <= 179):

        count_systolic_more_stage2 += 1

    elif (count >= 180):

        count_systolic_dangerous += 1



print(f"Number of data in the range Less (<120) is: {count_systolic_normal}")
print(50 * "==")
print(f"Number of data in the range between (120 to 129) is: {count_systolic_increased}")
print(50 * "==")
print(f"Number of data in the range between (130 to 139) is: {count_systolic_more_stage1}")
print(50 * "==")
print(f"Number of data in the range between (>=140) is: {count_systolic_more_stage2}")
print(50 * "==")
print(f"Number of data in the range more (>=180) is: {count_systolic_dangerous}")
```

```python
# ===============================================================================================  
count_sbp_female_normal = df[(df["bp.1s"] < 120) & (df["gender"] == "female")][["id"]]

# ===============================================================================================
count_sbp_female_increased = df[(df["bp.1s"].between(120, 129)) & (df["gender"] == "female")][["id"]]

# ===============================================================================================
count_sbp_female_more_stage1 = df[(df["bp.1s"].between(130, 139)) & (df["gender"] == "female")][["id"]]

# ===============================================================================================
count_sbp_female_more_stage2 = df[(df["bp.1s"].between(140, 179)) & (df["gender"] == "female")][["id"]]

# ===============================================================================================
count_sbp_female_dangerous = df[(df["bp.1s"] >= 180) & (df["gender"] == "female")][["id"]]





print(f"count sbp female normal: {count_sbp_female_normal.count()}")
print(50 * "==")
print(f"count sbp female increased: {count_sbp_female_increased.count()}")
print(50 * "==")
print(f"count sbp female more_stage1: {count_sbp_female_more_stage1.count()}")
print(50 * "==")
print(f"count sbp female more stage2: {count_sbp_female_more_stage2.count()}")
print(50 * "==")
print(f"count sbp female dangerous: {count_sbp_female_dangerous.count()}")
```

```python
count_normal = 0
count_more_stage_1 = 0
count_more_stage_2 = 0
count_dangerous = 0

for count in df["bp.1d"]:

    if (count < 80):

        count_normal += 1

    elif (80 <= count <= 89):

        count_more_stage_1 += 1

    elif (90 <= count <= 119):

        count_more_stage_2 += 1

    elif (count >= 120):

        count_dangerous += 1



print(f"Number of data in the range (<80): {count_normal}\n{50 * "=="}")
print(f"Number of data in the range (80 to 89): {count_more_stage_1}\n{50 * "=="}")
print(f"Number of data in the range (>=90): {count_more_stage_2}\n{50 * "=="}")
print(f"Number of data in the range (>=120): {count_dangerous}")
```

```python
# ============================================================================================================
count_dbp_female_normal = df[(df["bp.1d"] < 80) & (df["gender"] == "female")][["id"]]

# ============================================================================================================
count_dbp_female_more_stage_1 = df[(df["bp.1d"].between(80, 89)) & (df["gender"] == "female")][["id"]]

# ============================================================================================================
count_dbp_female_more_stage_2 = df[(df["bp.1d"].between(90, 119)) & (df["gender"] == "female")][["id"]]

# ============================================================================================================
count_dbp_female_dangerous = df[(df["bp.1d"] >= 120) & (df["gender"] == "female")][["id"]]






print(f"count DBP female normal: {count_dbp_female_normal.count()}")
print(50 * "==")
print(f"count DBP female more_stage_1: {count_dbp_female_more_stage_1.count()}")
print(50 * "==")
print(f"count DBP female more_stage_2: {count_dbp_female_more_stage_2.count()}")
print(50 * "==")
print(f"count DBP female dangerous: {count_dbp_female_dangerous.count()}")
```

```python
def check(dimension, testsize):

    r2 = 0.49487489442193666

    for column in clean_df:
        new_col_name = column + str(dimension)
        new_col_val = clean_df[column] **dimension
        clean_df.insert(0, new_col_name, new_col_val)
        x_train, x_test, y_train, y_test = train_test_split(feature_normal, y, test_size=testsize, random_state=0)
        regressor = LinearRegression()
        regressor.fit(x_train, y_train)
        y_pred = regressor.predict(x_test)
        r2_new = metrics.r2_score(y_test, y_pred)
        if r2_new <= r2:
            clean_df.drop([new_col_name], axis=1, inplace=True)
        else:
            r2 = r2_new

    print("R2 score: ", r2)




check(2, 0.2)
```
